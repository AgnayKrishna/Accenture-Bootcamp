# 🚀 Day 15 Challenge: Startup in a Box

> *"Every great company started as a crazy idea on a napkin. Your job: build the machine that turns napkins into Demo Day pitches."*

## 📖 The Story

**Moonshot Ventures**, a (fictional) venture studio, receives hundreds of wild product ideas every week. Their analysts can't keep up, so they've hired **you** as their founding AI engineer.

Your mission is to build an **AI venture studio**: a multi-agent system that takes **one wild idea** and turns it into a complete **🚀 Launch Kit**. The kit covers brand, market research, product plan, unit economics, a marketing campaign and an investor pitch, and it has to survive a tough **investor panel** at Demo Day.

You choose the framework. You design the team. You decide how the agents work together. 🧠

## 🎯 Rules of the Game

### ✅ Your system must…
1. **Use one agent framework** from Days 14–15 as its backbone: `smolagents`, `LangGraph`, `LlamaIndex`, `Google ADK` or `CrewAI`.
2. **Have at least 3 agents** with clearly different roles.
3. **Combine at least 2 agentic patterns** of your choice, and explain why in your design doc.
4. **Let agents call at least 2 tools**, from the Startup Toolkit or your own. The LLM should decide when to call them; don't call them yourself.
5. **Produce a `launch_kit` dictionary** that passes the Mission Check (section 7).
6. **Run top to bottom** after *Restart Kernel → Run All*.

### 🎨 You are free to…
- pick an idea from the deck **or invent your own**,
- design any architecture: a pipeline, a graph, a hierarchy, a debate, a swarm…,
- choose the agents, roles, prompts, tools and patterns, and how many iterations they get,
- add extra sections to the launch kit (logo prompt, landing page, customer interviews, risk register…).

### 🤝 Fair play
- The agents must generate the content. No hard-coding the answers!
- Keep a full run under **~50 LLM calls**, because we all share one Azure OpenAI quota.
- Never print, hard-code or commit API keys.

## 📦 The Launch Kit (your deliverable)

Your system must produce a Python `dict` called **`launch_kit`** with the keys below. Values can be strings, lists, dicts or Pydantic models; the format inside each section is up to you.

| Key | What it should contain |
|---|---|
| `startup_name` | A catchy name you've checked is available |
| `one_liner` | A one-sentence elevator pitch |
| `target_customers` | Who exactly it's for (personas/segments) |
| `market_insights` | Market size, growth, trends and risks |
| `competitors` | Main competitors and how you'll beat them |
| `mvp_features` | A prioritised list of MVP features |
| `pricing` | Pricing model and price points |
| `unit_economics` | LTV, CAC, LTV/CAC and payback, calculated with the tool |
| `marketing_campaign` | Slogan, launch channels and a few social posts |
| `investor_pitch` | Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script) |
| `investor_feedback` | What the investor panel said (scores, verdicts, what changed) |

Extra keys are welcome. `show_launch_kit()` renders them too, and it will even display a `landing_page_html` section as a web page. 🌐

import litellm
from google.adk.agents import LlmAgent, LoopAgent, ParallelAgent, SequentialAgent

**Useful:** the workflow agents `SequentialAgent`, `ParallelAgent` and `LoopAgent`. `output_key` saves an agent's answer in session state, and `{market_report}` in another agent's instruction reads it back. `sub_agents=[...]` lets the LLM delegate.

<summary><b>👥 CrewAI</b> (use a fresh environment)</summary>

```python
from crewai import LLM, Agent, Crew, Process, Task
from crewai.tools import tool

llm = LLM(model=f"openai/{AZURE_OPENAI_DEPLOYMENT}", base_url=AZURE_OPENAI_ENDPOINT,
          api_key=os.environ["AZURE_OPENAI_API_KEY"])
market_tool = tool("Market Data")(get_market_data)

analyst = Agent(role="Market Analyst", goal="Size up startup markets", backstory="You love data.",
                tools=[market_tool], llm=llm)
task = Task(description="Summarise the {sector} market.", expected_output="3 bullet points", agent=analyst)
crew = Crew(agents=[analyst], tasks=[task], process=Process.sequential)
result = await crew.kickoff_async(inputs={"sector": "pets"})  # in notebooks, always use kickoff_async
print(result)
```

**Useful:** `context=[...]` chains tasks, `async_execution=True` runs tasks in parallel, `Process.hierarchical` with `manager_llm=llm` adds a manager agent, and `output_pydantic=MyPydanticModel` gives structured outputs.
</details>

````python
# 🧰 The Startup Toolkit: fictional mock data + framework-agnostic tools.
# Every tool is a plain Python function with type hints and a docstring,
# so you can wrap it with any framework's tool helper (see the cheat sheet).
import hashlib
import json
import random
import re

from IPython.display import HTML, Markdown, display

# ---------------------------------------------------------------------------
# 🃏 The Wild Idea Deck
# ---------------------------------------------------------------------------
WILD_IDEA_DECK = [
    {"id": 1, "sector": "home",
     "idea": "An on-demand app that matches travelling plant owners with vetted 'plant sitters' who water, talk to, and send selfies of your plants."},
    {"id": 2, "sector": "entertainment",
     "idea": "An AI dream journal that turns parents' dreams into personalised bedtime stories for their kids."},
    {"id": 3, "sector": "pets",
     "idea": "Noise-cancelling headphones for dogs that switch on automatically during fireworks and thunderstorms."},
    {"id": 4, "sector": "food",
     "idea": "A smart lunchbox and companion app that turns eating vegetables into a game for picky kids."},
    {"id": 5, "sector": "education",
     "idea": "A language-learning app where you practise by debating AI versions of historical figures."},
    {"id": 6, "sector": "social",
     "idea": "'Grandparents-as-a-Service': matching retirees with busy families for storytelling, homework help and cooking lessons."},
    {"id": 7, "sector": "fitness",
     "idea": "Gym equipment that turns your workout into electricity and credits it to your home energy bill."},
    {"id": 8, "sector": "sustainability",
     "idea": "Carbon-negative coffee pods grown from mushroom mycelium that you can plant in your garden after use."},
    {"id": 9, "sector": "mobility",
     "idea": "A marketplace where office buildings rent their empty parking spots to food trucks at lunchtime."},
    {"id": 10, "sector": "entertainment",
     "idea": "A dating app for board-game lovers where every first date is a cooperative game night."},
]

# ---------------------------------------------------------------------------
# 📊 Fictional market data (numbers are made up for this exercise!)
# ---------------------------------------------------------------------------
MARKET_DATA = {
    "home": {"market_size_usd_bn": 48, "annual_growth_pct": 7.5, "avg_spend_usd_per_month": 25,
             "trends": ["Urban apartment gardening boom", "Gig-economy home services", "Smart-home sensors"],
             "risks": ["Trust & safety for in-home visits", "Seasonal demand (holidays)", "Low willingness to pay"],
             "benchmarks": {"typical_cac_usd": 30, "typical_monthly_churn_pct": 8}},
    "entertainment": {"market_size_usd_bn": 210, "annual_growth_pct": 9.2, "avg_spend_usd_per_month": 12,
                      "trends": ["Generative-AI personalised content", "Niche communities over mass platforms", "Offline-first experiences"],
                      "risks": ["Content moderation", "Subscription fatigue", "Fast-moving copycats"],
                      "benchmarks": {"typical_cac_usd": 18, "typical_monthly_churn_pct": 9}},
    "pets": {"market_size_usd_bn": 320, "annual_growth_pct": 6.1, "avg_spend_usd_per_month": 120,
             "trends": ["Pet humanisation ('fur babies')", "Premium wellness & anxiety products", "Subscription boxes"],
             "risks": ["Animal-welfare and vet claims", "Hardware returns", "Seasonal spikes (fireworks season)"],
             "benchmarks": {"typical_cac_usd": 35, "typical_monthly_churn_pct": 6}},
    "food": {"market_size_usd_bn": 95, "annual_growth_pct": 5.4, "avg_spend_usd_per_month": 40,
             "trends": ["Healthy eating for kids", "Gamified nutrition", "Connected kitchen devices"],
             "risks": ["Food-safety regulations", "Parents' screen-time concerns", "Hardware margins"],
             "benchmarks": {"typical_cac_usd": 40, "typical_monthly_churn_pct": 7}},
    "education": {"market_size_usd_bn": 165, "annual_growth_pct": 12.8, "avg_spend_usd_per_month": 15,
                  "trends": ["AI tutors and conversation practice", "Gamification", "Micro-learning"],
                  "risks": ["Low completion rates", "Historical accuracy & bias", "Free alternatives"],
                  "benchmarks": {"typical_cac_usd": 22, "typical_monthly_churn_pct": 10}},
    "social": {"market_size_usd_bn": 60, "annual_growth_pct": 8.0, "avg_spend_usd_per_month": 80,
               "trends": ["Loneliness economy", "Intergenerational connection", "Care marketplaces"],
               "risks": ["Safeguarding & background checks", "Two-sided marketplace cold start", "Insurance"],
               "benchmarks": {"typical_cac_usd": 55, "typical_monthly_churn_pct": 5}},
    "fitness": {"market_size_usd_bn": 105, "annual_growth_pct": 6.9, "avg_spend_usd_per_month": 45,
                "trends": ["Gamified workouts", "Sustainability-minded consumers", "Connected equipment"],
                "risks": ["Expensive hardware", "Energy regulations", "Motivation drop-off after January"],
                "benchmarks": {"typical_cac_usd": 60, "typical_monthly_churn_pct": 6}},
    "sustainability": {"market_size_usd_bn": 140, "annual_growth_pct": 14.5, "avg_spend_usd_per_month": 30,
                       "trends": ["Carbon-negative products", "Bio-materials (mycelium, algae)", "Circular economy"],
                       "risks": ["Greenwashing accusations", "Manufacturing scale-up", "Premium pricing"],
                       "benchmarks": {"typical_cac_usd": 45, "typical_monthly_churn_pct": 4}},
    "mobility": {"market_size_usd_bn": 75, "annual_growth_pct": 10.3, "avg_spend_usd_per_month": 200,
                 "trends": ["Under-used urban space", "Food-truck culture", "B2B marketplaces"],
                 "risks": ["City permits & zoning", "Liability", "Chicken-and-egg marketplace dynamics"],
                 "benchmarks": {"typical_cac_usd": 150, "typical_monthly_churn_pct": 3}},
}

# ---------------------------------------------------------------------------
# 🏢 Fictional competitor database
# ---------------------------------------------------------------------------
COMPETITORS = [
    {"name": "Leafy Nanny", "sector": "home", "tagline": "Plant-care tips from a chatbot", "keywords": ["plants", "plant care", "gardening"], "pricing": "$4.99/month", "weakness": "No in-person service"},
    {"name": "GreenThumb Robotics", "sector": "home", "tagline": "Self-watering smart pots", "keywords": ["plants", "smart pot", "watering", "iot"], "pricing": "$89 one-time", "weakness": "Only works with their own pots"},
    {"name": "SnoozeStory", "sector": "entertainment", "tagline": "AI bedtime stories on demand", "keywords": ["bedtime", "stories", "kids", "ai", "parents"], "pricing": "$7.99/month", "weakness": "Generic stories, nothing from the family's own life"},
    {"name": "DreamLog", "sector": "entertainment", "tagline": "Track and analyse your dreams", "keywords": ["dream", "journal", "sleep"], "pricing": "Free / $3.99 pro", "weakness": "Analysis only, no creative output"},
    {"name": "Roll for Love", "sector": "entertainment", "tagline": "Dating for video gamers", "keywords": ["dating", "games", "gamers"], "pricing": "Freemium, $14.99/month", "weakness": "Video games only, no tabletop community"},
    {"name": "StormHug", "sector": "pets", "tagline": "Anxiety-calming vests for dogs", "keywords": ["dog", "anxiety", "fireworks", "thunder"], "pricing": "$45 one-time", "weakness": "Doesn't address noise itself"},
    {"name": "Barkly Box", "sector": "pets", "tagline": "Monthly toy & treat box for dogs", "keywords": ["dog", "subscription", "toys", "treats"], "pricing": "$29/month", "weakness": "Commodity product, high churn"},
    {"name": "VeggieQuest", "sector": "food", "tagline": "Nutrition game app for kids", "keywords": ["kids", "vegetables", "game", "nutrition"], "pricing": "$5.99/month", "weakness": "Parents worry about extra screen time"},
    {"name": "LunchBotics", "sector": "food", "tagline": "Temperature-controlled smart bento", "keywords": ["lunchbox", "smart", "kids", "bento"], "pricing": "$129 one-time", "weakness": "Expensive, no engagement features"},
    {"name": "Lingo Legends", "sector": "education", "tagline": "Learn languages through quests", "keywords": ["language", "learning", "gamification"], "pricing": "Freemium, $9.99/month", "weakness": "Scripted dialogues, no real conversation"},
    {"name": "ChronoChat", "sector": "education", "tagline": "Chat with historical figures", "keywords": ["history", "historical figures", "chat", "ai"], "pricing": "$9.99/month", "weakness": "Trivia-focused, not designed for language practice"},
    {"name": "GrandBuddies", "sector": "social", "tagline": "Companion visits for seniors", "keywords": ["seniors", "retirees", "companionship", "elderly"], "pricing": "$25/hour", "weakness": "Treats seniors as recipients, not contributors"},
    {"name": "Village Nanny", "sector": "social", "tagline": "Babysitter marketplace", "keywords": ["childcare", "families", "babysitting", "marketplace"], "pricing": "15% booking fee", "weakness": "Transactional, no long-term relationships"},
    {"name": "PowerPedal", "sector": "fitness", "tagline": "Bike desks that charge your devices", "keywords": ["electricity", "energy", "bike", "workout"], "pricing": "$399 one-time", "weakness": "Tiny energy output, novelty factor"},
    {"name": "FitCoin Gym", "sector": "fitness", "tagline": "Gym that rewards workouts with points", "keywords": ["gym", "rewards", "workout", "points"], "pricing": "$35/month", "weakness": "Points have little real-world value"},
    {"name": "ShroomPack", "sector": "sustainability", "tagline": "Mycelium packaging for brands", "keywords": ["mycelium", "mushroom", "packaging", "b2b"], "pricing": "B2B contracts", "weakness": "No consumer brand"},
    {"name": "EcoPod Co.", "sector": "sustainability", "tagline": "Compostable coffee pods", "keywords": ["coffee", "pods", "compostable", "capsules"], "pricing": "$0.60/pod", "weakness": "Compostable is not carbon-negative"},
    {"name": "SpotSwap", "sector": "mobility", "tagline": "Peer-to-peer parking rental", "keywords": ["parking", "rental", "marketplace", "cars"], "pricing": "20% commission", "weakness": "Cars only, no commercial vehicles"},
    {"name": "TruckTrack", "sector": "mobility", "tagline": "Find food trucks near you", "keywords": ["food truck", "locator", "lunch"], "pricing": "Free (ads)", "weakness": "Consumer app only, doesn't solve where trucks can park"},
]

# ---------------------------------------------------------------------------
# 🦈 The Demo Day investor panel (fictional)
# ---------------------------------------------------------------------------
INVESTOR_PANEL = [
    {"name": "Ada Okafor", "nickname": "The Analyst", "fund": "Spreadsheet Capital",
     "loves": ["big markets", "healthy unit economics (LTV/CAC >= 3)", "defensible moats"],
     "hates": ["hand-wavy numbers", "tiny niches"],
     "catchphrase": "Show me the numbers."},
    {"name": "Rex Moon", "nickname": "The Visionary", "fund": "Moonshot Ventures",
     "loves": ["bold visions", "great storytelling", "creating new categories"],
     "hates": ["me-too products", "boring names"],
     "catchphrase": "Will this change how people live?"},
    {"name": "Priya Shah", "nickname": "The Operator", "fund": "Shipyard Partners",
     "loves": ["a tight MVP", "a clear go-to-market plan", "realistic milestones"],
     "hates": ["feature bloat", "vague target customers"],
     "catchphrase": "What ships in the first 90 days?"},
]

_TAKEN_NAMES = {"nova", "spark", "zen", "pulse", "flow", "bloom", "apex", "orbit", "nexus", "hive", "leaf", "dream"}


def draw_wild_idea(seed: int | None = None) -> dict:
    """Draw a random idea card from the Wild Idea Deck (pass a seed to make it repeatable)."""
    return random.Random(seed).choice(WILD_IDEA_DECK)


# ---------------------------------------------------------------------------
# 🔧 Tools (give these to your agents!)
# ---------------------------------------------------------------------------
def get_market_data(sector: str) -> str:
    """Get market size, growth, trends, risks and benchmark CAC/churn for a market sector.

    Args:
        sector: Market sector, one of: home, entertainment, pets, food, education, social, fitness, sustainability, mobility.

    Returns:
        JSON string with the (fictional) market data for the sector.
    """
    key = sector.strip().lower()
    if key not in MARKET_DATA:
        return json.dumps({"error": f"Unknown sector '{sector}'.", "available_sectors": sorted(MARKET_DATA)})
    return json.dumps({"sector": key, **MARKET_DATA[key]})


def search_competitors(keyword: str) -> str:
    """Search the competitor database for companies matching a keyword (e.g. 'dog', 'parking', 'coffee').

    Args:
        keyword: A single keyword or short phrase describing the product or customer.

    Returns:
        JSON string with up to 5 matching competitors, including their pricing and main weakness.
    """
    needle = keyword.strip().lower()
    matches = [
        c for c in COMPETITORS
        if needle and (needle in c["name"].lower() or needle in c["tagline"].lower()
                       or needle == c["sector"] or any(needle in k for k in c["keywords"]))
    ]
    if not matches:
        return json.dumps({"keyword": keyword, "competitors": [], "note": "No direct competitors found. Blue ocean, or no market?"})
    return json.dumps({"keyword": keyword, "competitors": matches[:5]})


def check_name_availability(name: str) -> str:
    """Check whether a startup name is available as a .com/.ai domain and whether it has trademark risk.

    Args:
        name: The proposed startup name, e.g. 'BarkBuds'.

    Returns:
        JSON string with domain availability, trademark risk and a verdict.
    """
    slug = re.sub(r"[^a-z0-9]", "", name.lower())
    if not slug:
        return json.dumps({"name": name, "error": "Name must contain letters or digits."})
    digest = int(hashlib.sha256(slug.encode()).hexdigest(), 16)
    clash = any(taken in slug for taken in _TAKEN_NAMES)
    com_free = not clash and digest % 3 != 0
    ai_free = not clash and digest % 4 != 0
    risk = "high" if clash else ("medium" if digest % 7 == 0 else "low")
    usable = (com_free or ai_free) and risk != "high"
    return json.dumps({
        "name": name,
        "domain_com": f"{slug}.com is " + ("available" if com_free else "taken"),
        "domain_ai": f"{slug}.ai is " + ("available" if ai_free else "taken"),
        "trademark_risk": risk,
        "verdict": "Looks usable!" if usable else "Pick another name.",
    })


def estimate_unit_economics(price_per_month_usd: float, cost_to_serve_per_month_usd: float,
                            acquisition_cost_usd: float, monthly_churn_pct: float) -> str:
    """Calculate customer lifetime value (LTV), LTV/CAC ratio and CAC payback period.

    For one-off purchases, spread the price over the expected product life to get a monthly figure.

    Args:
        price_per_month_usd: Average monthly revenue per customer in USD.
        cost_to_serve_per_month_usd: Average monthly cost to serve one customer in USD.
        acquisition_cost_usd: Customer acquisition cost (CAC) in USD.
        monthly_churn_pct: Percentage of customers lost each month, e.g. 5 for 5%.

    Returns:
        JSON string with gross margin, LTV, LTV/CAC, payback months and a verdict.
    """
    margin = price_per_month_usd - cost_to_serve_per_month_usd
    churn = max(monthly_churn_pct, 0.1) / 100
    lifetime_months = 1 / churn
    ltv = margin * lifetime_months
    ltv_cac = ltv / acquisition_cost_usd if acquisition_cost_usd > 0 else float("inf")
    payback = acquisition_cost_usd / margin if margin > 0 else float("inf")
    if margin <= 0:
        verdict = "Negative margin: you lose money on every customer."
    elif ltv_cac >= 3:
        verdict = "Healthy: investors like LTV/CAC >= 3."
    elif ltv_cac >= 1:
        verdict = "Risky: profitable per customer, but growth will be expensive."
    else:
        verdict = "Unsustainable: each customer costs more than they bring in."
    return json.dumps({
        "gross_margin_per_month_usd": round(margin, 2),
        "gross_margin_pct": round(100 * margin / price_per_month_usd, 1) if price_per_month_usd else None,
        "expected_lifetime_months": round(lifetime_months, 1),
        "ltv_usd": round(ltv, 2),
        "ltv_to_cac": round(ltv_cac, 2) if ltv_cac != float("inf") else "infinite",
        "cac_payback_months": round(payback, 1) if payback != float("inf") else "never",
        "verdict": verdict,
    })


def get_investor_panel() -> str:
    """Get the profiles of the Demo Day investors: what each one loves, hates and asks about.

    Returns:
        JSON string with the investor profiles.
    """
    return json.dumps(INVESTOR_PANEL)


# ---------------------------------------------------------------------------
# 📦 Launch Kit contract, checker and viewer
# ---------------------------------------------------------------------------
LAUNCH_KIT_SPEC = {
    "startup_name": "A catchy name you've checked is available",
    "one_liner": "A one-sentence elevator pitch",
    "target_customers": "Who exactly it's for (personas/segments)",
    "market_insights": "Market size, growth, trends and risks",
    "competitors": "Main competitors and how you'll beat them",
    "mvp_features": "A prioritised list of MVP features",
    "pricing": "Pricing model and price points",
    "unit_economics": "LTV, CAC, LTV/CAC and payback, calculated with the tool",
    "marketing_campaign": "Slogan, launch channels and a few social posts",
    "investor_pitch": "Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script)",
    "investor_feedback": "What the investor panel said (scores, verdicts, what changed)",
}


def _to_plain(value):
    """Convert Pydantic models (and containers of them) into plain Python data."""
    if hasattr(value, "model_dump"):
        return _to_plain(value.model_dump())
    if isinstance(value, dict):
        return {k: _to_plain(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [_to_plain(v) for v in value]
    return value


def _is_filled(value) -> bool:
    if value is None:
        return False
    if isinstance(value, str):
        return bool(value.strip())
    if isinstance(value, (list, tuple, dict, set)):
        return len(value) > 0
    return bool(str(value).strip())


def check_launch_kit(kit) -> bool:
    """Mission control check: verify that the launch kit contains every required section."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict):
        print(f"❌ launch_kit must be a dict (or a Pydantic model), got {type(kit).__name__}.")
        return False
    passed = 0
    for key, description in LAUNCH_KIT_SPEC.items():
        ok = _is_filled(kit.get(key))
        passed += ok
        print(f"{'✅' if ok else '❌'} {key:<20}" + ("" if ok else f" <- missing: {description}"))
    extras = [k for k in kit if k not in LAUNCH_KIT_SPEC]
    if extras:
        print(f"\n✨ Extra sections (nice!): {', '.join(extras)}")
    print(f"\nLaunch readiness: {passed}/{len(LAUNCH_KIT_SPEC)}")
    if passed == len(LAUNCH_KIT_SPEC):
        print("🚀 All systems go! Your startup is ready for Demo Day.")
    return passed == len(LAUNCH_KIT_SPEC)


def _render(value, indent: int = 0) -> str:
    pad = "  " * indent
    if isinstance(value, dict):
        lines = []
        for k, v in value.items():
            if isinstance(v, (dict, list)):
                lines.append(f"{pad}- **{k}**:\n{_render(v, indent + 1)}")
            else:
                lines.append(f"{pad}- **{k}**: {v}")
        return "\n".join(lines)
    if isinstance(value, list):
        return "\n".join(
            f"{pad}-\n{_render(v, indent + 1)}" if isinstance(v, (dict, list)) else f"{pad}- {v}" for v in value
        )
    return f"{pad}{value}"


def show_launch_kit(kit) -> None:
    """Render the launch kit as a nicely formatted Markdown report."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict) or not kit:
        print("Nothing to show yet. Build your launch_kit first!")
        return
    md = [f"# 🚀 {kit.get('startup_name', 'Unnamed Startup')}"]
    if kit.get("one_liner"):
        md.append(f"> {kit['one_liner']}")
    for key, value in kit.items():
        if key in ("startup_name", "one_liner", "landing_page_html") or not _is_filled(value):
            continue
        md.append(f"## {key.replace('_', ' ').title()}\n\n{_render(value)}")
    display(Markdown("\n\n".join(md)))
    if kit.get("landing_page_html"):
        display(HTML(kit["landing_page_html"]))


print(f"🧰 Startup Toolkit loaded: {len(WILD_IDEA_DECK)} wild ideas, {len(MARKET_DATA)} sectors, "
      f"{len(COMPETITORS)} competitors, {len(INVESTOR_PANEL)} investors.")
````

- **Evaluator-optimizer:** the investor panel scores the pitch, the founder rewrites it using their biggest concerns, and this repeats until the average is 8 or more, or three rounds have passed.
- **Tool use (ReAct):** the market, competitor, brand and finance agents decide on their own when and how often to call their tools.

**Stop conditions and safety limits:** a call tracker stops the run at 45 LLM calls (the brief says about 50). The pitch loop runs at most 3 rounds. **Tips from the Moonshot engineering team:**
- 🐣 **Start small:** get one agent with one tool working end-to-end, then add agents one by one.

- 🛑 **Always cap loops** with `max_steps`, a maximum number of rounds, or `max_iterations`.

- 📐 **Structured outputs** (Pydantic models) make assembling the `launch_kit` much easier.

````python
# 5.1 Create the model, a call tracker and the tools
import html
import json
import operator
import re
import threading
import time
from collections import defaultdict
from typing import Annotated, TypedDict

from langchain.agents import create_agent
from langchain.agents.structured_output import ToolStrategy
from langchain_core.callbacks import BaseCallbackHandler
from langchain_core.messages import AIMessage, ToolMessage
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.graph import END, START, StateGraph
from pydantic import BaseModel, Field

MAX_LLM_CALLS = 45     # hard stop, the brief asks for roughly 50 calls at most
MAX_PITCH_ROUNDS = 3   # the investors get three looks at the pitch, no more
TARGET_SCORE = 8.0     # average investor score that ends the loop


class CallTracker(BaseCallbackHandler):
    """Counts LLM calls, tokens and time per stage, and stops the run if it runs away."""

    raise_error = True  # let the RuntimeError below really stop the graph

    def __init__(self):
        self._lock = threading.Lock()
        self.reset()

    def reset(self):
        self.calls = defaultdict(int)
        self.tokens_in = defaultdict(int)
        self.tokens_out = defaultdict(int)
        self.seconds = defaultdict(float)
        self._started = {}

    @property
    def total_calls(self):
        return sum(self.calls.values())

    @staticmethod
    def _stage(tags):
        for t in tags or []:
            if t.startswith("stage:"):
                return t[len("stage:"):]
        return "other"

    def on_chat_model_start(self, serialized, messages, *, run_id, tags=None, **kwargs):
        with self._lock:
            if self.total_calls >= MAX_LLM_CALLS:
                raise RuntimeError(f"Stopped: more than {MAX_LLM_CALLS} LLM calls in one run.")
            stage = self._stage(tags)
            self.calls[stage] += 1
            self._started[run_id] = (stage, time.time())

    def on_llm_end(self, response, *, run_id, **kwargs):
        with self._lock:
            stage, t0 = self._started.pop(run_id, ("other", time.time()))
            self.seconds[stage] += time.time() - t0
            try:
                usage = response.generations[0][0].message.usage_metadata or {}
            except Exception:
                usage = {}
            self.tokens_in[stage] += usage.get("input_tokens", 0)
            self.tokens_out[stage] += usage.get("output_tokens", 0)

    def on_llm_error(self, error, *, run_id, **kwargs):
        with self._lock:
            self._started.pop(run_id, None)


tracker = CallTracker()

# No temperature / max_tokens here: GPT-5 models only accept the defaults
llm = ChatOpenAI(
    model=AZURE_OPENAI_DEPLOYMENT,
    base_url=AZURE_OPENAI_ENDPOINT,
    api_key=os.environ["AZURE_OPENAI_API_KEY"],
    callbacks=[tracker],
    max_retries=5,
    timeout=120,
)

# Wrap the toolkit functions so the agents can call them
market_tool = tool(get_market_data)
competitor_tool = tool(search_competitors)
name_tool = tool(check_name_availability)
economics_tool = tool(estimate_unit_economics)
TOOL_NAMES = {"get_market_data", "search_competitors", "check_name_availability", "estimate_unit_economics"}


# Structured outputs: every agent that writes a launch-kit section returns one of these
class Persona(BaseModel):
    name: str = Field(description="Short persona name, for example 'Anxious dog-parent Anna'")
    who: str = Field(description="Age, situation and where they live, in one sentence")
    pain_point: str
    why_they_would_pay: str


class CustomerProfiles(BaseModel):
    personas: list[Persona] = Field(description="Exactly three personas")
    primary_segment: str = Field(description="The one segment we launch to first, in one sentence")


class BrandChoice(BaseModel):
    startup_name: str
    one_liner: str = Field(description="One sentence elevator pitch, under 25 words")
    names_checked: list[str] = Field(description="Every candidate name you checked with the tool")
    why_this_name: str


class MvpFeature(BaseModel):
    priority: str = Field(description="P0 (needed at launch), P1 (soon after) or P2 (later)")
    feature: str
    why: str


class ProductPlan(BaseModel):
    mvp_features: list[MvpFeature] = Field(description="5 to 7 features, most important first")
    pricing_model: str
    price_points: list[str] = Field(description="For example 'Device: $129 one-time'")
    monthly_equivalent_price_usd: float = Field(description="Revenue per customer per month (device price spread over 24 months plus any subscription)")
    cost_to_serve_per_month_usd: float = Field(description="Cost per customer per month (hardware cost spread over 24 months plus service costs)")
    ninety_day_milestones: list[str] = Field(description="Three milestones for the first 90 days")


class Campaign(BaseModel):
    slogan: str = Field(description="Under 8 words")
    launch_channels: list[str] = Field(description="Three channels, each with a one-line reason")
    social_posts: list[str] = Field(description="Three ready-to-publish posts with hashtags")


class Slide(BaseModel):
    title: str
    content: str = Field(description="At most 60 words")


class Pitch(BaseModel):
    slides: list[Slide] = Field(description="Six slides: Problem, Solution, Market, Business model and unit economics, Go-to-market, The ask")
    the_ask: str = Field(description="One sentence: how much we raise and what it buys")
    changes_from_last_round: str = Field(default="", description="What changed after the investor feedback. Leave empty for the first draft")


class Review(BaseModel):
    score: int = Field(description="Whole number from 1 to 10")
    verdict: str = Field(description="One sentence verdict")
    biggest_concern: str
    fix: str = Field(description="The single change that would raise your score the most")
````

````python
# 5.2 Create the agents: roles, instructions, tools
def text_of(message):
    """Plain text of a chat message (some models return a list of content blocks)."""
    content = message.content
    if isinstance(content, list):
        return "".join(b.get("text", "") if isinstance(b, dict) else str(b) for b in content)
    return str(content)


# --- Agents that decide for themselves when to call a tool (create_agent = ReAct loop) ---
market_analyst = create_agent(
    llm,
    tools=[market_tool],
    name="market_analyst",
    system_prompt=(
        "You are the Market Analyst at a venture studio. The founders need facts, not vibes.\n"
        "Always call get_market_data for the startup's sector first, and take every number from what "
        "the tool returns. Never invent figures.\n"
        "Then write a market report of at most 150 words with these labelled parts: Size and growth, "
        "Trends, Risks, Benchmarks (typical CAC and monthly churn, copied exactly from the tool). "
        "End with one line starting 'So what for this startup:'."
    ),
)

competitor_analyst = create_agent(
    llm,
    tools=[competitor_tool],
    name="competitor_analyst",
    system_prompt=(
        "You are the Competitor Analyst at a venture studio.\n"
        "Call search_competitors two or three times with different short keywords that fit the idea "
        "(the customer, the problem, the kind of product). Only talk about competitors the tool returns. "
        "If nothing relevant comes back, say so honestly.\n"
        "Write at most 170 words: one line per relevant competitor (name, pricing, main weakness), then "
        "'How we beat them:' with two or three concrete points."
    ),
)

brand_agent = create_agent(
    llm,
    tools=[name_tool],
    name="brand_strategist",
    system_prompt=(
        "You are the Brand Strategist at a venture studio.\n"
        "Invent three distinctive candidate names (one or two words, easy to say, nothing generic like "
        "Nova or Spark). Check EVERY candidate with check_name_availability. Pick one whose verdict says "
        "it looks usable. Then give the final choice: startup_name, a one_liner, the names you checked "
        "and a short reason for the pick."
    ),
    response_format=ToolStrategy(BrandChoice),
)

finance_analyst = create_agent(
    llm,
    tools=[economics_tool],
    name="finance_analyst",
    system_prompt=(
        "You are the Finance Analyst at a venture studio. You never work out LTV, LTV/CAC or payback "
        "in your head: the estimate_unit_economics tool does the maths.\n"
        "Call it twice: (1) a base case using the product plan's monthly-equivalent price and cost to "
        "serve plus the CAC and monthly churn benchmarks from the market report, and (2) a cautious case "
        "with CAC 30% higher and churn 2 points higher.\n"
        "Then write at most 120 words: both cases with the numbers the tool returned, what the verdict "
        "means for this startup, and one thing that would improve LTV/CAC."
    ),
)


# --- Writers: one structured LLM call each, no tools needed ---
def structured(model_class):
    return llm.with_structured_output(model_class, method="function_calling")


customer_profiler = structured(CustomerProfiles)
product_manager = structured(ProductPlan)
marketer = structured(Campaign)
pitch_writer = structured(Pitch)

CUSTOMER_PROMPT = (
    "You are the Customer Researcher at a venture studio. Create exactly three realistic customer "
    "personas for the idea. Be concrete (age, situation, city type), never 'everyone'. Name the "
    "segment we should launch to first."
)
PRODUCT_PROMPT = (
    "You are the Product Manager at a venture studio. Plan an MVP that can ship in 90 days: 5 to 7 "
    "features tagged P0/P1/P2 (P0 only for what is needed on launch day), a pricing model that fits the "
    "market data, and three milestones for the first 90 days.\n"
    "If the product is hardware, give the device price AND a monthly equivalent by spreading the price "
    "over 24 months (plus any subscription). Cost to serve per month = hardware cost spread over the "
    "same 24 months plus service costs per customer. Keep numbers realistic for the market data."
)
MARKETING_PROMPT = (
    "You are the Head of Marketing at a venture studio. Write a slogan (under 8 words), three launch "
    "channels with a one-line reason each, and three social posts ready to publish with hashtags, all "
    "in the voice of the brand. Use only facts from the plan, no invented statistics."
)
PITCH_PROMPT = (
    "You are the founder preparing a Demo Day pitch. Write exactly six slides: Problem, Solution, "
    "Market, Business model and unit economics, Go-to-market, The ask. Each slide has at most 60 words. "
    "Quote market and unit-economics numbers exactly as they appear in the inputs and never invent "
    "figures. This is a pre-seed pitch, so do not claim results you do not have: show the plan to get "
    "proof instead (what you will test, with how many users, and the target you need to hit). The panel "
    "will include an analyst who wants numbers, a visionary who wants a bold story and an operator who "
    "wants a clear plan for the first 90 days."
)
LANDING_PROMPT = (
    "You are a web designer. Write ONE self-contained HTML page (inline CSS, no JavaScript, no external "
    "files or images) that works as the landing page of the startup: hero with name, slogan and one "
    "liner, three feature cards, the price, and a sign-up button. Keep it under 90 lines and use only "
    "facts from the launch kit. Return only the HTML."
)


# --- The investor panel: one agent per investor, each with their own taste ---
def investor_system_prompt(p):
    return (
        f"You are {p['name']}, '{p['nickname']}' at {p['fund']}, judging a startup pitch at Demo Day.\n"
        f"You love: {', '.join(p['loves'])}. You hate: {', '.join(p['hates'])}. "
        f"Your catchphrase: \"{p['catchphrase']}\"\n"
        "This is a pre-seed pitch: there is no product, no customers and no pilot data yet, so judge the "
        "quality of the plan, the market logic and the numbers you are given, not traction that cannot "
        "exist yet. Reward a credible, measurable plan to get proof within 90 days.\n"
        "Be tough but fair and judge only what is in the pitch and the numbers. A first draft rarely "
        "deserves more than a 7. Give 8 or more only if the pitch really answers what you care about. "
        "Score from 1 to 10."
    )


investors = {
    p["name"]: {"profile": p, "brain": structured(Review), "system": investor_system_prompt(p)}
    for p in INVESTOR_PANEL
}


# --- Small helpers shared by the graph nodes ---
def run_agent(agent, prompt, stage):
    """Run a tool-using agent. The tag lets the tracker count its LLM calls per stage."""
    return agent.invoke(
        {"messages": [{"role": "user", "content": prompt}]},
        config={"tags": [f"stage:{stage}"], "recursion_limit": 14},
    )


def ask(brain, system, user, stage, attempts=2):
    """One structured LLM call, retried once if the model returns something unparsable."""
    for attempt in range(attempts):
        try:
            return brain.invoke([("system", system), ("human", user)], config={"tags": [f"stage:{stage}"]})
        except Exception as err:
            if attempt == attempts - 1:
                raise
            print(f"   (retrying {stage}: {type(err).__name__})")


def tool_calls_in(messages, stage):
    """Pair every tool call the model made with the result the tool returned."""
    results = {m.tool_call_id: m.content for m in messages if isinstance(m, ToolMessage)}
    calls = []
    for m in messages:
        if isinstance(m, AIMessage):
            for c in m.tool_calls:
                if c["name"] in TOOL_NAMES:
                    calls.append({"stage": stage, "tool": c["name"], "args": c["args"], "result": results.get(c["id"])})
    return calls


def as_json(value):
    try:
        return json.loads(value)
    except (TypeError, ValueError):
        return value


def warn_if_unused(calls, tool_name, stage):
    if not any(c["tool"] == tool_name for c in calls):
        print(f"   ⚠️ the {stage} agent never called {tool_name}")


def pitch_to_text(pitch):
    lines = [f"{i}. {s['title']}: {s['content']}" for i, s in enumerate(pitch["slides"], 1)]
    return "\n".join(lines + [f"The ask: {pitch['the_ask']}"])
````

````python
# 5.3 Wire up the workflow
#
#   START -> market_analyst ---+
#         -> competitor_analyst +--> brand_agent -> product_manager -> finance_analyst -> marketer -> pitch_writer
#         -> customer_profiler -+                                                                     |  |  |
#                                                                          ada <----------------------+  |  +----> priya
#                                                                                               rex <----+
#   (three investors score in parallel) -> tally -> average >= 8 or round 3?  no -> back to pitch_writer
#                                                                              yes -> landing_page -> END

def merge_lists(old, new):
    return (old or []) + (new or [])


class StudioState(TypedDict, total=False):
    idea: str
    sector: str
    market_report: str
    competitor_report: str
    customers: dict
    brand: dict
    product: dict
    unit_economics: dict
    campaign: dict
    pitch: dict
    round: int
    avg_score: float
    history: list
    landing_page_html: str
    reviews: Annotated[list, merge_lists]         # investors write at the same time
    tool_log: Annotated[list, merge_lists]        # several agents log tool calls at the same time
    pitch_versions: Annotated[list, merge_lists]


# ---------- Stage 1: research, three agents in parallel ----------
def market_node(state):
    print("🔎 Market analyst is researching...")
    res = run_agent(market_analyst, f"Startup idea: {state['idea']}\nSector: {state['sector']}", "market")
    calls = tool_calls_in(res["messages"], "market")
    warn_if_unused(calls, "get_market_data", "market")
    return {"market_report": text_of(res["messages"][-1]), "tool_log": calls}


def competitor_node(state):
    print("🕵️ Competitor analyst is searching...")
    res = run_agent(competitor_analyst, f"Startup idea: {state['idea']}\nSector: {state['sector']}", "competitors")
    calls = tool_calls_in(res["messages"], "competitors")
    warn_if_unused(calls, "search_competitors", "competitor")
    return {"competitor_report": text_of(res["messages"][-1]), "tool_log": calls}


def customer_node(state):
    print("🧑‍🤝‍🧑 Customer researcher is building personas...")
    out = ask(customer_profiler, CUSTOMER_PROMPT, f"Startup idea: {state['idea']}\nSector: {state['sector']}", "customers")
    return {"customers": out.model_dump()}


# ---------- Stage 2: the chain ----------
def usable(check):
    return isinstance(check, dict) and check.get("verdict") == "Looks usable!"


def brand_node(state):
    print("🏷️ Brand strategist is naming the startup...")
    prompt = (
        f"Startup idea: {state['idea']}\nLaunch segment: {state['customers']['primary_segment']}\n"
        f"Competitors to stay away from (do not sound like them):\n{state['competitor_report'][:600]}"
    )
    all_calls = []
    for attempt in range(2):
        res = run_agent(brand_agent, prompt, "brand")
        calls = tool_calls_in(res["messages"], "brand")
        all_calls += calls
        choice = res["structured_response"]
        slug = re.sub(r"[^a-z0-9]", "", choice.startup_name.lower())
        check = None
        for c in calls:  # find what the tool said about the name that was picked
            if c["tool"] == "check_name_availability" and re.sub(r"[^a-z0-9]", "", c["args"].get("name", "").lower()) == slug:
                check = as_json(c["result"])
        if usable(check):
            break
        print(f"   ⚠️ '{choice.startup_name}' did not pass the availability check, asking for another name")
        prompt += f"\nYour previous pick '{choice.startup_name}' was not confirmed as available. Pick a different name and check it."
    brand = choice.model_dump()
    brand["availability_check"] = check if check else "not checked"
    return {"brand": brand, "tool_log": all_calls}


def product_node(state):
    print("📦 Product manager is planning the MVP...")
    user = (
        f"Startup: {state['brand']['startup_name']} - {state['brand']['one_liner']}\nIdea: {state['idea']}\n\n"
        f"Customers: {json.dumps(state['customers'])}\n\nMarket report:\n{state['market_report']}\n\n"
        f"Competitors:\n{state['competitor_report']}"
    )
    return {"product": ask(product_manager, PRODUCT_PROMPT, user, "product").model_dump()}


def finance_node(state):
    print("🧮 Finance analyst is running the numbers...")
    prompt = (
        f"Startup: {state['brand']['startup_name']}\nProduct plan: {json.dumps(state['product'])}\n\n"
        f"Market report (use its benchmarks):\n{state['market_report']}"
    )
    res = run_agent(finance_analyst, prompt, "finance")
    calls = tool_calls_in(res["messages"], "finance")
    warn_if_unused(calls, "estimate_unit_economics", "finance")
    scenarios = [{"inputs": c["args"], "result": as_json(c["result"])} for c in calls if c["tool"] == "estimate_unit_economics"]
    return {"unit_economics": {"scenarios": scenarios, "analysis": text_of(res["messages"][-1])}, "tool_log": calls}


def marketing_node(state):
    print("📣 Head of marketing is writing the campaign...")
    user = (
        f"Startup: {state['brand']['startup_name']} - {state['brand']['one_liner']}\n"
        f"Customers: {json.dumps(state['customers'])}\nProduct plan: {json.dumps(state['product'])}"
    )
    return {"campaign": ask(marketer, MARKETING_PROMPT, user, "marketing").model_dump()}


# ---------- Stage 3: pitch, then the investor panel (parallel) with a feedback loop ----------
def pitch_node(state):
    rnd = state.get("round", 0)
    facts = (
        f"Startup: {state['brand']['startup_name']} - {state['brand']['one_liner']}\n"
        f"Customers: {json.dumps(state['customers'])}\nMarket report:\n{state['market_report']}\n"
        f"Competitors:\n{state['competitor_report']}\nProduct plan: {json.dumps(state['product'])}\n"
        f"Unit economics (from the finance tool):\n{json.dumps(state['unit_economics'])}\n"
        f"Campaign: {json.dumps(state['campaign'])}"
    )
    if rnd == 0:
        print("🎤 Founder is writing the first pitch...")
        user = facts
    else:
        print(f"🎤 Founder is revising the pitch (round {rnd + 1})...")
        feedback = "\n".join(
            f"- {r['investor']} ({r['score']}/10): {r['biggest_concern']} Suggested fix: {r['fix']}"
            for r in state["reviews"] if r["round"] == rnd
        )
        user = (
            f"{facts}\n\nYour previous pitch:\n{pitch_to_text(state['pitch'])}\n\n"
            f"Investor feedback (average {state['avg_score']:.1f}/10):\n{feedback}\n\n"
            "Revise the pitch so that it answers every concern, and describe what you changed in "
            "changes_from_last_round."
        )
    pitch = ask(pitch_writer, PITCH_PROMPT, user, "pitch").model_dump()
    return {"pitch": pitch, "round": rnd + 1, "pitch_versions": [pitch]}


def make_investor_node(name):
    info = investors[name]

    def node(state):
        print(f"🦈 {name} ({info['profile']['nickname']}) is scoring round {state['round']}...")
        user = (
            f"Idea: {state['idea']}\n\nThe pitch:\n{pitch_to_text(state['pitch'])}\n\n"
            f"Unit economics from the finance tool:\n{json.dumps(state['unit_economics']['scenarios'])}\n\n"
            f"Market report:\n{state['market_report']}"
        )
        review = ask(info["brain"], info["system"], user, "investors").model_dump()
        review["score"] = max(1, min(10, int(review["score"])))
        review.update({"investor": name, "nickname": info["profile"]["nickname"], "round": state["round"]})
        return {"reviews": [review]}

    return node


def tally_node(state):
    rnd = state["round"]
    this_round = [r for r in state["reviews"] if r["round"] == rnd]
    avg = sum(r["score"] for r in this_round) / len(this_round)
    print(f"📊 Round {rnd}: average score {avg:.1f} ({', '.join(str(r['score']) for r in this_round)})")
    entry = {"round": rnd, "average": round(avg, 2), "scores": {r["investor"]: r["score"] for r in this_round}}
    return {"avg_score": avg, "history": (state.get("history") or []) + [entry]}


def route_after_tally(state):
    if state["avg_score"] >= TARGET_SCORE:
        print("✅ The panel is happy.")
        return "landing_page"
    if state["round"] >= MAX_PITCH_ROUNDS:
        print(f"🛑 Stopping after {MAX_PITCH_ROUNDS} rounds.")
        return "landing_page"
    return "pitch_writer"


def landing_node(state):
    print("🌐 Web designer is building the landing page...")
    facts = {
        "name": state["brand"]["startup_name"], "one_liner": state["brand"]["one_liner"],
        "slogan": state["campaign"]["slogan"], "features": [f["feature"] for f in state["product"]["mvp_features"][:3]],
        "price_points": state["product"]["price_points"],
    }
    out = llm.invoke([("system", LANDING_PROMPT), ("human", json.dumps(facts))], config={"tags": ["stage:landing"]})
    page = text_of(out).strip()
    page = re.sub(r"^```(?:html)?\s*|\s*```$", "", page).strip()
    return {"landing_page_html": page}


# ---------- Build the graph ----------
builder = StateGraph(StudioState)
builder.add_node("market_analyst", market_node)
builder.add_node("competitor_analyst", competitor_node)
builder.add_node("customer_profiler", customer_node)
builder.add_node("brand_agent", brand_node)
builder.add_node("product_manager", product_node)
builder.add_node("finance_analyst", finance_node)
builder.add_node("marketer", marketing_node)
builder.add_node("pitch_writer", pitch_node)
builder.add_node("tally", tally_node)
builder.add_node("landing_page", landing_node)

investor_nodes = []
for investor_name in investors:
    node_id = "investor_" + investor_name.split()[0].lower()
    builder.add_node(node_id, make_investor_node(investor_name))
    investor_nodes.append(node_id)

# parallel research, then wait for all three
for first in ("market_analyst", "competitor_analyst", "customer_profiler"):
    builder.add_edge(START, first)
builder.add_edge(["market_analyst", "competitor_analyst", "customer_profiler"], "brand_agent")

# the chain
builder.add_edge("brand_agent", "product_manager")
builder.add_edge("product_manager", "finance_analyst")
builder.add_edge("finance_analyst", "marketer")
builder.add_edge("marketer", "pitch_writer")

# investors score in parallel, then the evaluator-optimizer loop decides what happens next
for node_id in investor_nodes:
    builder.add_edge("pitch_writer", node_id)
builder.add_edge(investor_nodes, "tally")
builder.add_conditional_edges("tally", route_after_tally, {"pitch_writer": "pitch_writer", "landing_page": "landing_page"})
builder.add_edge("landing_page", END)

studio = builder.compile()
print("✅ Venture studio is wired up:", ", ".join(studio.get_graph().nodes))
````

Run your venture studio on `MY_IDEA` and collect the results into the `launch_kit` dictionary. You can fill it straight from your agents' outputs, or have a final "assembler" agent build it.

````python
# Run the venture studio on MY_IDEA and build the launch kit
def build_launch_kit(state, wall_seconds):
    """Collect what the agents produced into the launch_kit dict (no new content is written here)."""
    last_round = state["round"]
    final_reviews = [r for r in state["reviews"] if r["round"] == last_round]
    return {
        "startup_name": state["brand"]["startup_name"],
        "one_liner": state["brand"]["one_liner"],
        "target_customers": state["customers"],
        "market_insights": state["market_report"],
        "competitors": state["competitor_report"],
        "mvp_features": state["product"]["mvp_features"],
        "pricing": {
            "model": state["product"]["pricing_model"],
            "price_points": state["product"]["price_points"],
            "monthly_equivalent_price_usd": state["product"]["monthly_equivalent_price_usd"],
        },
        "unit_economics": state["unit_economics"],
        "marketing_campaign": state["campaign"],
        "investor_pitch": state["pitch"],
        "investor_feedback": {
            "rounds": state["history"],
            "final_average": round(state["avg_score"], 2),
            "reached_target": state["avg_score"] >= TARGET_SCORE,
            "final_reviews": final_reviews,
            "what_changed": [
                {"version": i, "changes": v["changes_from_last_round"]}
                for i, v in enumerate(state["pitch_versions"], 1) if v["changes_from_last_round"]
            ],
        },
        # extra sections
        "name_check": state["brand"]["availability_check"],
        "landing_page_html": state["landing_page_html"],
        "tool_calls": [{"stage": c["stage"], "tool": c["tool"], "args": c["args"]} for c in state["tool_log"]],
        "run_stats": {
            "llm_calls": tracker.total_calls,
            "seconds": round(wall_seconds, 1),
            "calls_per_stage": dict(tracker.calls),
        },
    }


tracker.reset()
started = time.time()
final_state = studio.invoke(
    {"idea": MY_IDEA["idea"], "sector": MY_IDEA["sector"], "round": 0, "reviews": [], "tool_log": [], "pitch_versions": []},
    config={"recursion_limit": 40},
)
launch_kit = build_launch_kit(final_state, time.time() - started)
print(f"\n🚀 Launch kit ready: {launch_kit['startup_name']} ({tracker.total_calls} LLM calls, {launch_kit['run_stats']['seconds']}s)")
````
